```python
# Scatter plot game
import matplotlib.pyplot as plt
import random
import numpy as np

fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(0, 50)
ax.set_ylim(0, 50)
ax.set_xticks([])
ax.set_yticks([])

player_pos = np.array([5, 5])
target_pos = np.array([random.randint(0, 50), random.randint(0, 50)])

player_scatter = ax.scatter(player_pos[0], player_pos[1], color='red')
target_scatter = ax.scatter(target_pos[0], target_pos[1], color='blue')

def on_key_press(event):
    global player_pos
    if event.key == 'up':
        player_pos[1] += 1
    elif event.key == 'down':
        player_pos[1] -= 1
    elif event.key == 'left':
        player_pos[0] -= 1
    elif event.key == 'right':
        player_pos[0] += 1

player_scatter.set_offsets(player_pos)
fig.canvas.draw_idle()

if np.linalg.norm(player_pos - target_pos) < 1:
    print("You win!")
    plt.close(fig)

fig.canvas.mpl_connect('key_press_event', on_key_press)

plt.legend()
plt.show()

```

```python
# Algebra practice game

import random

def algebra_game():
  score=0
  num_questions = 5

  for i in range(num_questions):
    a = random.randint(1, 10)
    b = random.randint(1, 10)
    operator = random.choice(['+', '-', '*', '/'])

    if operator == '+':
      answer = a + b
    elif operator == '-':
      answer = a - b
    elif operator == '*':
      answer = a * b
    elif operator == '/':
      answer = a / b

    print(f"Question {i+1}: {a} {operator} {b} = ?")
    user_answer = float(input())

    if user_answer == answer:
      print("Correct!")

      score += 1
    else:
      print(f"Incorrect. The correct answer is {answer}.")

  print(f"Your final score is {score}/{num_questions}.")

algebra_game()


```

```python
import math
import random
import numpy as np
import matplotlib.pyplot as plt

# Create the figure
fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(0, 80)
ax.set_ylim(0, 50)
ax.set_xlabel("X position")
ax.set_ylabel("Y position")
ax.set_xticks([])
ax.set_yticks([])
ax.set_title("Projectile Game")

# Random target position
target_x = random.randint(40, 75)
target_y = random.randint(10, 40)
ax.scatter(target_x, target_y, color='blue', s=100, label='Target ')

# Projectile variables
initial_speed = random.randint(10, 20)
angle = random.randint(20, 70)  # realistic angle
gravity = 9.8

# Convert angle to radians
angle_rad = math.radians(angle)

# Velocity components
velocity_x = initial_speed * math.cos(angle_rad)
velocity_y = initial_speed * math.sin(angle_rad)

# Time array
total_time = (2 * velocity_y) / gravity
t_values = np.linspace(0, total_time, 200)

# Projectile positions
x_values = velocity_x * t_values
y_values = velocity_y * t_values - 0.5 * gravity * t_values**2

# Draw the full trajectory
ax.plot(x_values, y_values, 'r-', linewidth=2, label='Trajectory ')
ax.scatter(x_values[-1], y_values[-1], color='red', s=50, label='Projectile End')

# Optional: check if projectile hits target
if any(np.sqrt((x_values - target_x)**2 + (y_values - target_y)**2) < 2):
    hit_text = "Target Hit! "
else:
    hit_text = "Missed!"

# Show hit/miss on the plot
ax.text(5, 45, hit_text, fontsize=14, color='green' if "Hit" in hit_text else 'red')

ax.legend()
plt.show()
```
